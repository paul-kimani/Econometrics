# Goodness of Fit

- Here, we measure how well a statistical model fits a set of observations.
  Does the model:
  1. Fit the data well?
  2. Does it explain the variations around the mean of $y$?
$$
TSS = \sum_{i=1}^{n} (y_i - \bar{y})^2
$$
This is the total sum of squares, which measures the total variation in the response variable $y$. i.e TSS quantifies how much the observed values of $y$ deviate from their mean $\bar{y}$.
Without the square, it is the total sum of deviations, which would be zero due to positive and negative deviations canceling each other out.

$$
RSS = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2
$$
This is the residual sum of squares (RSS), which measures the variation in $y$ that is not explained by the model. It quantifies how much the observed values of $y$ deviate from the predicted values $\hat{y}_i$ produced by the model.

$$
ESS = \sum_{i=1}^{n} (\hat{y}_i - \bar{y})^2
$$
This on the other hand is the explained sum of squares (ESS), which measures the variation in $y$ that is explained by the model. It quantifies how much the predicted values $\hat{y}_i$ deviate from the mean $\bar{y}$.(ESS)

We measure the ESS using the coefficient of determination, denoted as $R^2$:
$$
R^2 = \frac{ESS}{TSS} = 1 - \frac{RSS}{TSS} = 1 - \frac{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{n} (y_i - \bar{y})^2}
$$

For a sample, we include an n-1 in the folrmula to account for degrees of freedom:
$$
R^2 = \frac{ESS/(n-1)}{TSS/(n-1)} = 1 - \frac{RSS/(n-p)}{TSS/(n-1)} = 1 - \frac{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2/(n-p)}{\sum_{i=1}^{n} (y_i - \bar{y})^2/(n-1)}
$$

we realise that:
$$
0 \leq R^2 \leq 1
$$

The higher our $R^2$, the better our model fits the data. An $R^2$ of 1 indicates a perfect fit, while an $R^2$ of 0 indicates that the model does not explain any of the variation in the response variable.
For example in the Housing dataset, if we have an $R^2$ of 0.85, it means that 85% of the variation in house prices can be explained by the features in our model, while the remaining 15% is due to other factors not included in the model.

We can also say, relative to someone who predicts $Y$ with its mean $\bar{Y}$, a person who predicts $Y$ with our model is 85% times better(Accurate).

$$
TOTAL = EXPLAINED + UNEXPLAINED
$$

$$
TSS = ESS + RSS
$$


$$
\frac{TSS}{TSS} = \frac{ESS}{TSS} + \frac{RSS}{TSS}
$$

$$
1 = R^2 + \frac{RSS}{TSS} = R^2 + \frac{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{n} (y_i - \bar{y})^2}
$$
$\therefore$
$$
 R^2 = 1 - \frac{RSS}{TSS} = 1 - \frac{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{n} (y_i - \bar{y})^2}
$$

On further understanding of $R^2$, we find that it is equivalent to the square of the correlation between the observed values and the predicted values from the model.
Also, we realize that it has a weakness: i.e..,
It goes up with the addition of the addition of more predictors, even if those predictors do not significantly improve the model's fit/explanatory power.


**To address this weakness, we use the Adjusted R-squared, which adjusts for the number of predictors in the model:**
This $R^2_{adj}$ provides a more accurate measure of model fit by penalizing the addition of unnecessary predictors, hence it only goes up if the new predictor improves the model more than would be expected by chance.


It is calculated as follows:
$$
\bar{R}^2 = 1 - \left(1 - R^2\right) \frac{n - 1}{n - p} = 1 - \left(\frac{RSS/(n - p)}{TSS/(n - 1)}\right) = 1 - \frac{\sum_{i=1}^{n} (y_i - \hat{y}_i)^2/(n - p)}{\sum_{i=1}^{n} (y_i - \bar{y})^2/(n - 1)}
$$

Note that:
This adjusted $R^2$ can be negative if the model is a poor fit, which indicates that the model performs worse than simply predicting the mean of the response variable.
#### Proposition
observe that the ratios $\frac{1}{n-p}$ and $\frac{1}{n-1}$ are used to account for degrees of freedom in the model, and simplify to:
$$
\frac{1}{n-p} \times \frac{n-1}{1} = \frac{n-1}{n-p}
$$

Hence as p increases, the denominator $(n-p)$ decreases, leading to a larger adjustment factor $\frac{n-1}{n-p}$. i.e..,
as we add more predictors, the penalty for adding unnecessary predictors becomes more significant, which can decrease the adjusted $R^2$ if the new predictors do not improve the model sufficiently.
Mathematically:
as $p \to n$, $\frac{n-1}{n-p} \to \infty$.
Hence the whole equation tends to 1

   $$
   \beta = (X'X)^{-1} X'Y = (X'X)^{-1} X'(X\beta + U) = \beta + (X'X)^{-1} X'U
   $$

   $$

   $$

   $$
   the expectations:
   $$

   $$
We realise that even in the presence of heteroskedasticity, the OLS estimator $\hat{\beta}$ remains unbiased.

## So, does $\beta_{ols}$ still have the minimum variance property among all linear unbiased estimators?
We know that the variance of $\beta_{ols}$ is given by:


$$
Var(\beta_{ols}) = E[(\beta_{ols} - \beta)(\beta_{ols} - \beta)'] = E[((X'X)^{-1} X'U)((X'X)^{-1} X'U)'] = (X'X)^{-1} X' E[UU'] X (X'X)^{-1}
$$

$$
Var(\beta_{ols}) = (X'X)^{-1} X' E[UU'] X (X'X)^{-1} = (X'X)^{-1} X' \psi X (X'X)^{-1}
$$



under heteroskedasticity, variance of $\beta_{ols}$ is no longer simply $\sigma^2 (X'X)^{-1}$, but instead depends on the structure of $\psi$.....it is inflated!!!

- $var(\beta_{ols})$ is no longer the minimum variance among all linear unbiased estimators....is inflated

### is there a solution to heteroskedasticity?
Let $P'P = \psi^{-1}$. 
where P'P is the Cholesky decomposition of $\psi^{-1}$.
The intuitive idea here is to transform the original model by pre-multiplying both sides by the matrix P. where P is chosen such that it "whitens" the error terms, making them homoscedastic.
P is a matrix that when multiplied by the error terms U, transforms them to have constant variance. so we call P the whitening matrix.
$$
(P'P)^{-1} = \psi
$$
if we make the left hand side $I_n$, we shall aslo make the right hand side $I_n$.
$$
(p'p)^{-1} = \psi \Rightarrow P^{-1} P'^{-1} = \psi
$$
premultiply LHS with P and postmultiply RHS with P':
$$
P P^{-1} P'^{-1} P' = P \psi P'
$$
We realize that:
$$ 
P P^{-1} = I_n
$$
hence:
$$
I_n = P \psi P'
$$
So :
A solution/transformation exists to convert $\psi$ into $I_n$.
We have a transformation that can help recover the minimum variance property of the $\beta_{ols}$ estimator.
Its easy to see that for the matrix $\psi$
$$
\psi = \begin{bmatrix}
\lambda_1 & 0 & ... & 0 \\
0 & \lambda_2 & ... & 0 \\
\dots & \dots & \dots & \dots \\
0 & 0 & ... & \lambda_n \\
\end{bmatrix}
$$
Infact the matrix P can be constructed as:
$$
p = \begin{bmatrix}
\frac{1}{\sqrt{\lambda_1}} & 0 & ... & 0 \\
0 & \frac{1}{\sqrt{\lambda_2}} & ... & 0 \\
0 & 0 & ... & \frac{1}{\sqrt{\lambda_n}} \\
\dots & \dots & \dots & \dots \\
0 & 0 & ... & \frac{1}{\sqrt{\lambda_n}} \\
\end{bmatrix}
$$
So for the model $ Y = X\beta + U$
The transformation to retrieve the minimum variance property is:`
$$
\frac{Y}{\sqrt{\lambda_i}} = \frac{X}{\sqrt{\lambda_i}} \beta + \frac{U}{\sqrt{\lambda_i}}
$$
Hence we have transformed the original model into a new model with homoscedastic errors. and can say:
$$
Y^* = X^* \beta + U^*
$$
and we can apply OLS to this transformed model to obtain efficient estimates of $\beta$.
$$
\hat{\beta}_{GLS} = (X^{*'} X^*)^{-1} X^{*'} Y^*
$$
where $\hat{\beta}_{GLS}$ is the Generalized Least Squares (GLS) estimator, which is efficient in the presence of heteroskedasticity.
Since $p$ is known in our case, we call it the Feasible Generalized Least Squares (FGLS) estimator.
$$
\hat{\beta}_{FGLS} = (X^{*'} X^*)^{-1} X^{*'} Y^* = (X' P' P X)^{-1} X' P' P Y
$$
What is the purpose of $\beta_{FGLS}$?
It helps us trace back the minimum variance property of $\beta_{ols}$ in the presence of heteroskedasticity and donates it to the $\beta_{OLS}$ for the purposes of testing hypothesis and constructing confidence intervals.

## **(a) Show that this model is heteroscedastic**

We have:

$$
y_i = \beta_1 + \beta_2 x_{2i} + \varepsilon_i
$$

$$
\varepsilon_i = \sqrt{x_{2i}} \cdot u_i, \quad u_i \overset{iid}{\sim} N(0,1)
$$

The conditional variance of \(\varepsilon_i\) given \(x_{2i}\) is:

$$
\mathrm{Var}(\varepsilon_i | x_{2i}) = \mathrm{Var}(\sqrt{x_{2i}} \cdot u_i | x_{2i})
$$

$$
= (\sqrt{x_{2i}})^2 \cdot \mathrm{Var}(u_i | x_{2i})
$$
$$
= x_{2i} \cdot 1
$$
since \(u_i\) is independent of \(x_{2i}\) and \(\mathrm{Var}(u_i) = 1\).

Thus:

$$
\mathrm{Var}(\varepsilon_i | x_{2i}) = x_{2i}.
$$

This depends on \(x_{2i}\), so the model is **heteroscedastic**.

$$
\boxed{\text{Heteroscedastic because Var}(\varepsilon_i|x_{2i}) = x_{2i} \neq \text{constant}}
$$

### Alternatively:

The model is given by:
$$y_i = \beta_1 + \beta_2 x_{2i} + \epsilon_i$$
The error term $\epsilon_i$ is defined by:
$$\epsilon_i = \sqrt{x_{2i}} u_i$$
where $u_i \sim \text{IID } N(0, 1)$ and $x_{2i}$ is a non-stochastic variable.


## (a) Show that this model is heteroscedastic.

A model is **heteroscedastic** if the variance of the error term $\epsilon_i$, conditional on the independent variable $x_{2i}$, is **not constant** across observations $i$.

The conditional variance of $\epsilon_i$ is:
$$\text{Var}(\epsilon_i | x_{2i}) = E[(\epsilon_i - E[\epsilon_i])^2 | x_{2i}]$$

### 1. Find the conditional mean of $\epsilon_i$:
$$E[\epsilon_i | x_{2i}] = E[\sqrt{x_{2i}} u_i | x_{2i}]$$
Since $x_{2i}$ is non-stochastic (or fixed when conditioning) and $E[u_i] = 0$:
$$E[\epsilon_i | x_{2i}] = \sqrt{x_{2i}} E[u_i] = \sqrt{x_{2i}} \cdot 0 = 0$$

### 2. Find the conditional variance of $\epsilon_i$:
$$\text{Var}(\epsilon_i | x_{2i}) = E[(\epsilon_i - 0)^2 | x_{2i}] = E[\epsilon_i^2 | x_{2i}]$$
Substitute $\epsilon_i = \sqrt{x_{2i}} u_i$:
$$\text{Var}(\epsilon_i | x_{2i}) = E[(\sqrt{x_{2i}} u_i)^2 | x_{2i}] = E[x_{2i} u_i^2 | x_{2i}]$$
Since $x_{2i}$ is non-stochastic and $\text{Var}(u_i) = E[u_i^2] = 1$:
$$\text{Var}(\epsilon_i | x_{2i}) = x_{2i} E[u_i^2] = x_{2i} \cdot 1 = x_{2i}$$

Since $\text{Var}(\epsilon_i | x_{2i}) = x_{2i}$, and the variance **depends on $x_{2i}$**, it is **not constant** across all observations $i$ (unless $x_{2i}$ is constant, which is generally not the case).
Therefore, the model is **heteroscedastic**.

## (b) Estimate $\beta_1$ and $\beta_2$ by OLS.

The given empirical information is:
* $y = [4, 3, -5, -6]'$
* $x_2 = [2, 1, -1, -2]'$
* The model is $y_i = \beta_1 + \beta_2 x_{2i} + \epsilon_i$.

We need to use Ordinary Least Squares (OLS) formulas:
$$\hat{\beta}_2^{\text{OLS}} = \frac{\sum_{i=1}^n (x_{2i} - \bar{x}_2)(y_i - \bar{y})}{\sum_{i=1}^n (x_{2i} - \bar{x}_2)^2}$$
$$\hat{\beta}_1^{\text{OLS}} = \bar{y} - \hat{\beta}_2^{\text{OLS}} \bar{x}_2$$

### 1. Calculate the means:
$$\bar{y} = \frac{4 + 3 + (-5) + (-6)}{4} = \frac{-4}{4} = -1$$
$$\bar{x}_2 = \frac{2 + 1 + (-1) + (-2)}{4} = \frac{0}{4} = 0$$

### 2. Calculate the components:
Since $\bar{x}_2 = 0$, the numerator simplifies to $\sum x_{2i} y_i$ and the denominator to $\sum x_{2i}^2$.

* **Numerator $\sum x_{2i} y_i$:**
    $$(2)(4) + (1)(3) + (-1)(-5) + (-2)(-6) = 8 + 3 + 5 + 12 = 28$$

* **Denominator $\sum x_{2i}^2$:**
    $$2^2 + 1^2 + (-1)^2 + (-2)^2 = 4 + 1 + 1 + 4 = 10$$

### 3. Calculate $\hat{\beta}_2^{\text{OLS}}$:
$$\hat{\beta}_2^{\text{OLS}} = \frac{28}{10} = 2.8$$

### 4. Calculate $\hat{\beta}_1^{\text{OLS}}$:
Since $\bar{x}_2 = 0$:
$$\hat{\beta}_1^{\text{OLS}} = \bar{y} - \hat{\beta}_2^{\text{OLS}} \bar{x}_2 = -1 - (2.8)(0) = -1$$

The OLS estimates are **$\hat{\beta}_1 = -1$** and **$\hat{\beta}_2 = 2.8$**.

## (c) Discuss the statistical properties of the OLS estimators in this case.

The model $y_i = \beta_1 + \beta_2 x_{2i} + \epsilon_i$ satisfies the classical assumptions for OLS, *except* the assumption of **homoscedasticity** (constant error variance), as shown in part (a).

The Gauss-Markov Theorem states that under the classical assumptions (including homoscedasticity), OLS estimators are **Best Linear Unbiased Estimators (BLUE)**.

In the presence of **heteroscedasticity**:

* **Unbiasedness and Consistency:** The OLS estimators $\hat{\beta}_1$ and $\hat{\beta}_2$ remain **unbiased** and **consistent**. This is because heteroscedasticity does not affect the assumptions required for unbiasedness, primarily $E[\epsilon_i | x_{2i}] = 0$.

* **Efficiency:** The OLS estimators are **no longer efficient**. They are not the **BLUE**. The existence of heteroscedasticity means there is a more efficient linear unbiased estimator (the Generalized Least Squares (GLS) estimator).

* **Variance Estimates and Inference:** The calculated standard errors and confidence intervals based on the standard OLS formula for the variance of $\hat{\beta}$ will be **biased** and **inconsistent**. This leads to **incorrect t-statistics, F-statistics**, and $\boldsymbol{p}$-values. Consequently, hypothesis testing and inference using the standard OLS output will be **invalid** or **misleading**.

## (d) Discuss how you would transform the data so that you could remove the heteroscedasticity.

The issue is that the variance of the error term depends on $x_{2i}$:
$$\text{Var}(\epsilon_i | x_{2i}) = \sigma_i^2 = x_{2i}$$

To remove heteroscedasticity and restore the assumption of homoscedasticity, we use **Weighted Least Squares (WLS)**, which is equivalent to transforming the model into one with a homoscedastic error term. This is a form of **Generalized Least Squares (GLS)**.

The transformation involves dividing every term in the original equation by $\sqrt{\sigma_i^2}$, which is the standard deviation of the error term, $\sigma_i = \sqrt{x_{2i}}$.

### 1. Original Model:
$$y_i = \beta_1 + \beta_2 x_{2i} + \epsilon_i$$

### 2. Transformation:
Divide the entire equation by $\sqrt{x_{2i}}$:
$$\frac{y_i}{\sqrt{x_{2i}}} = \beta_1 \frac{1}{\sqrt{x_{2i}}} + \beta_2 \frac{x_{2i}}{\sqrt{x_{2i}}} + \frac{\epsilon_i}{\sqrt{x_{2i}}}$$

### 3. Transformed Model:
Simplify the terms to define the new variables:
$$y_i^* = \beta_1 x_{1i}^* + \beta_2 x_{2i}^* + \epsilon_i^*$$
where:
* $y_i^* = \frac{y_i}{\sqrt{x_{2i}}}$
* $x_{1i}^* = \frac{1}{\sqrt{x_{2i}}}$ (The transformed intercept term)
* $x_{2i}^* = \frac{x_{2i}}{\sqrt{x_{2i}}} = \sqrt{x_{2i}}$
* $\epsilon_i^* = \frac{\epsilon_i}{\sqrt{x_{2i}}}$

### 4. Homoscedasticity Check for $\epsilon_i^*$:
The new error term is $\epsilon_i^* = \frac{\epsilon_i}{\sqrt{x_{2i}}}$. Since $\epsilon_i = \sqrt{x_{2i}} u_i$:
$$\epsilon_i^* = \frac{\sqrt{x_{2i}} u_i}{\sqrt{x_{2i}}} = u_i$$
Since $u_i \sim \text{IID } N(0, 1)$, the variance of the new error term is:
$$\text{Var}(\epsilon_i^*) = \text{Var}(u_i) = 1$$
Since the new variance is **1 (a constant)**, the transformed model is now **homoscedastic**. We would then apply OLS to the transformed data $(y^*, x_1^*, x_2^*)$ to obtain the **GLS (WLS)** estimates.

## (e) Now estimate the model with the empirical information given in section (b), but by GLS.

We use the transformation established in part (d), but first, we must note the **"Take due note of the minus"** hint.

The error variance is $\sigma_i^2 = x_{2i}$. However, looking at the data from part (b):
$$x_2 = [2, 1, \mathbf{-1}, \mathbf{-2}]'$$
The variance of a real-valued variable must be $\mathbf{\geq 0}$. Since $x_{2i}$ has negative values, $x_{2i}$ cannot be the variance $\sigma_i^2$.

Let's re-examine the implied heteroscedasticity structure, assuming the true variance is proportional to $|x_{2i}|$, and that $\sigma_i^2 = c|x_{2i}|$ for some constant $c$. Since $u_i \sim N(0, 1)$, $\text{Var}(\epsilon_i | x_{2i}) = x_{2i} \text{Var}(u_i) = x_{2i}$. This implies that the model statement is only valid if $x_{2i} > 0$ for all $i$.

The "minus" in $x_2$ and the structure $\sigma_i^2 = x_{2i}$ strongly suggests that we should use the **absolute value** of $x_{2i}$ as the variance structure:
$$\sigma_i^2 = |x_{2i}|$$
The transformation factor is thus $\sqrt{\sigma_i^2} = \sqrt{|x_{2i}|}$.

### 1. Calculate the Transformation Factor $\sqrt{|x_{2i}|}$:
$$\sqrt{|x_2|} = [\sqrt{|2|}, \sqrt{|1|}, \sqrt{|-1|}, \sqrt{|-2|}]' = [\sqrt{2}, 1, 1, \sqrt{2}]'$$
The values are approximately $[1.414, 1.000, 1.000, 1.414]'$.

### 2. Calculate Transformed Variables ($y^*, x_1^*, x_2^*$):
* $y_i^* = \frac{y_i}{\sqrt{|x_{2i}|}}$
* $x_{1i}^* = \frac{1}{\sqrt{|x_{2i}|}}$
* $x_{2i}^* = \frac{x_{2i}}{\sqrt{|x_{2i}|}} = \frac{\text{sgn}(x_{2i})|x_{2i}|}{\sqrt{|x_{2i}|}} = \text{sgn}(x_{2i})\sqrt{|x_{2i}|}$

| $i$ | $y_i$ | $x_{2i}$ | $\sqrt{|x_{2i}|}$ | $y_i^* = \frac{y_i}{\sqrt{|x_{2i}|}}$ | $x_{1i}^* = \frac{1}{\sqrt{|x_{2i}|}}$ | $x_{2i}^* = \frac{x_{2i}}{\sqrt{|x_{2i}|}}$ |
| :-: | :-: | :-: | :-: | :-: | :-: | :-: |
| 1 | 4 | 2 | $\sqrt{2}$ | $4/\sqrt{2} = 2\sqrt{2}$ | $1/\sqrt{2}$ | $2/\sqrt{2} = \sqrt{2}$ |
| 2 | 3 | 1 | 1 | $3/1 = 3$ | $1/1 = 1$ | $1/1 = 1$ |
| 3 | -5 | -1 | 1 | $-5/1 = -5$ | $1/1 = 1$ | $-1/1 = -1$ |
| 4 | -6 | -2 | $\sqrt{2}$ | $-6/\sqrt{2} = -3\sqrt{2}$ | $1/\sqrt{2}$ | $-2/\sqrt{2} = -\sqrt{2}$ |

We will now apply OLS to the transformed variables $\mathbf{y}^*$ on $\mathbf{X}^* = [\mathbf{x}_1^*, \mathbf{x}_2^*]$.

### 3. OLS on Transformed Data (GLS):
The matrix form is $\mathbf{y}^* = \mathbf{X}^* \boldsymbol{\beta} + \boldsymbol{\epsilon}^*$, and the OLS estimator is:
$$\hat{\boldsymbol{\beta}}^{\text{GLS}} = (\mathbf{X}^{*'} \mathbf{X}^*)^{-1} \mathbf{X}^{*'} \mathbf{y}^*$$

* **$\mathbf{X}^{*'} \mathbf{X}^*$:**
    $$\mathbf{X}^{*'} \mathbf{X}^* = \begin{pmatrix} \sum x_{1i}^{*2} & \sum x_{1i}^* x_{2i}^* \\ \sum x_{1i}^* x_{2i}^* & \sum x_{2i}^{*2} \end{pmatrix}$$
    * $\sum x_{1i}^{*2} = (1/\sqrt{2})^2 + 1^2 + 1^2 + (1/\sqrt{2})^2 = 1/2 + 1 + 1 + 1/2 = 3$
    * $\sum x_{2i}^{*2} = (\sqrt{2})^2 + 1^2 + (-1)^2 + (-\sqrt{2})^2 = 2 + 1 + 1 + 2 = 6$
    * $\sum x_{1i}^* x_{2i}^* = (1/\sqrt{2})(\sqrt{2}) + (1)(1) + (1)(-1) + (1/\sqrt{2})(-\sqrt{2}) = 1 + 1 - 1 - 1 = 0$
    $$\mathbf{X}^{*'} \mathbf{X}^* = \begin{pmatrix} 3 & 0 \\ 0 & 6 \end{pmatrix}$$
    Since the matrix is diagonal, $\mathbf{X}^{*'} \mathbf{X}^*$ is simpler:
    $$(\mathbf{X}^{*'} \mathbf{X}^*)^{-1} = \begin{pmatrix} 1/3 & 0 \\ 0 & 1/6 \end{pmatrix}$$

* **$\mathbf{X}^{*'} \mathbf{y}^*$:**
    $$\mathbf{X}^{*'} \mathbf{y}^* = \begin{pmatrix} \sum x_{1i}^* y_i^* \\ \sum x_{2i}^* y_i^* \end{pmatrix}$$
    * $\sum x_{1i}^* y_i^* = (1/\sqrt{2})(2\sqrt{2}) + (1)(3) + (1)(-5) + (1/\sqrt{2})(-3\sqrt{2}) = 2 + 3 - 5 - 3 = -3$
    * $\sum x_{2i}^* y_i^* = (\sqrt{2})(2\sqrt{2}) + (1)(3) + (-1)(-5) + (-\sqrt{2})(-3\sqrt{2}) = 4 + 3 + 5 + 6 = 18$
    $$\mathbf{X}^{*'} \mathbf{y}^* = \begin{pmatrix} -3 \\ 18 \end{pmatrix}$$

* **$\hat{\boldsymbol{\beta}}^{\text{GLS}}$:**
    $$\hat{\boldsymbol{\beta}}^{\text{GLS}} = \begin{pmatrix} \hat{\beta}_1 \\ \hat{\beta}_2 \end{pmatrix} = \begin{pmatrix} 1/3 & 0 \\ 0 & 1/6 \end{pmatrix} \begin{pmatrix} -3 \\ 18 \end{pmatrix} = \begin{pmatrix} (1/3)(-3) + 0 \\ 0 + (1/6)(18) \end{pmatrix} = \begin{pmatrix} -1 \\ 3 \end{pmatrix}$$

The GLS estimates are **$\hat{\beta}_1 = -1$** and **$\hat{\beta}_2 = 3$**.
